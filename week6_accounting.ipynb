```python
def calculate_balance(records):
  income=0
  expense=0
  for kind,amount in records:
    if kind=="收入":
      income=income+amount
    else:
      expense=expense+amount
  balance=income-expense
  return income,expense,balance

records=[]

while True:
  kind=input("這是收入還是支出?(收入/支出/結束)")
  if kind=="結束":
    break
  while True:
    text = input("金額?")
    if text.isdigit():
        amount = int(text)
        break
    else:
        print("請輸入數字")
  records.append((kind,amount))

income,expense,balance=calculate_balance(records)
print("總收入",income)
print("總支出",expense)
print("結餘",balance)

print("\n本日收支明細")
for kind,amount in records:
  print(kind,":",amount)
```
